```python
import matplotlib.pyplot as plt
import numpy as np
import scienceplots  # noqa: F401

plt.style.use("science")
```

```python
mclmc = np.load("../data/dc_mclmc.npz")
rw = np.load("../data/dc_mh.npz")
pathfinder = np.load("../data/dc_pathfinder.npz")

fig, (ax1, ax2, ax3) = plt.subplots(ncols=3, figsize=(20, 5))

hist_kwargs = dict(
    alpha=0.8,
    bins=50,
    density=True,
)

ax1.hist(mclmc["alpha"][2000:], label="MCLMC", **hist_kwargs)
ax1.hist(rw["alpha"][10_000:], label="RW", **hist_kwargs)
ax1.hist(pathfinder["alpha"], label="VI", **hist_kwargs)

ax2.hist(mclmc["kappa"][2000:], **hist_kwargs)
ax2.hist(rw["kappa"][10_000:], **hist_kwargs)
ax2.hist(pathfinder["kappa"], **hist_kwargs)

ax3.hist(mclmc["epsilon"][2000:], **hist_kwargs)
ax3.hist(rw["epsilon"][10_000:], **hist_kwargs)
ax3.hist(pathfinder["epsilon"], **hist_kwargs)

ax1.set_title(r"Distribution of $\alpha$")
ax2.set_title(r"Distribution of $\kappa_0$")
ax3.set_title(r"Distribution of $\epsilon_0$")

ax1.set_xlabel(r"$\alpha$")
ax2.set_xlabel(r"$\kappa_0$")

ax1.set_ylabel("Probability density")
ax2.set_ylabel("Probability density")
ax3.set_ylabel("Probability density")


handles, labels = ax1.get_legend_handles_labels()
fig.legend(
    handles,
    labels,
    loc="lower center",
    ncol=3,
    frameon=False,
)

fig.suptitle(
    "Posterior distributions obtained using Linear Sweep DC Voltammetry",
    fontsize=18,
    y=0.98,
)

plt.show()
```
